## Importar un módulo con `import`

````python
import math

resultado = math.sqrt(16)
print(resultado)  # 4.0
````

## Módulos de la biblioteca estándar

### `math` — operaciones matemáticas

````python
import math

print(math.sqrt(144))      # 12.0 — raíz cuadrada
print(math.pow(2, 10))     # 1024.0 — potencia
print(math.floor(3.9))     # 3 — redondeo hacia abajo
print(math.ceil(3.1))      # 4 — redondeo hacia arriba
print(math.pi)             # 3.141592653589793
print(abs(-7))             # 7 — abs() es función incorporada
````

### `random` — números y selecciones aleatorias

````python
import random

print(random.randint(1, 6))          # número entero entre 1 y 6
print(random.random())               # decimal entre 0.0 y 1.0

frutas = ["manzana", "pera", "uva", "mango"]
print(random.choice(frutas))         # elige uno al azar
print(random.sample(frutas, 2))      # elige 2 sin repetir
````

### `datetime` — fechas y horas

````python
from datetime import date, datetime

hoy = date.today()
print(hoy)
print(f"Año: {hoy.year}, Mes: {hoy.month}, Día: {hoy.day}")

ahora = datetime.now()
print(ahora.strftime("%d/%m/%Y %H:%M"))
````

### `os` — interactuar con el sistema operativo

````python
import os

print(f"Carpeta actual: {os.getcwd()}")
print(f"Archivos aquí: {os.listdir('.')}")
````

### Métodos de texto (no requieren importar nada)

````python
texto = "  Hola, Technovation Girls!  "

print(texto.strip())          # elimina espacios
print(texto.lower())          # minúsculas
print(texto.upper())          # mayúsculas
print(texto.replace("Hola", "Bienvenida"))
print(texto.split(","))       # divide en lista
print("Girls" in texto)       # True
````

## Ejemplo completo: generador de contraseñas

````python
import random
import string

def generar_contrasena(longitud=12, incluir_simbolos=True):
    letras = string.ascii_letters    # a-z A-Z
    numeros = string.digits          # 0-9
    simbolos = string.punctuation    # !@#$%...

    if incluir_simbolos:
        caracteres = letras + numeros + simbolos
    else:
        caracteres = letras + numeros

    contrasena = "".join(random.choices(caracteres, k=longitud))
    return contrasena

def evaluar_fortaleza(contrasena):
    tiene_mayuscula = any(c.isupper() for c in contrasena)
    tiene_minuscula = any(c.islower() for c in contrasena)
    tiene_numero = any(c.isdigit() for c in contrasena)
    tiene_simbolo = any(c in string.punctuation for c in contrasena)

    puntaje = sum([tiene_mayuscula, tiene_minuscula, tiene_numero, tiene_simbolo])

    if puntaje == 4:
        return "Muy fuerte"
    elif puntaje == 3:
        return "Fuerte"
    elif puntaje == 2:
        return "Media"
    else:
        return "Debil"

# Programa principal
print("--- Generador de contrasenas ---\n")

for i in range(5):
    contrasena = generar_contrasena(longitud=16)
    fortaleza = evaluar_fortaleza(contrasena)
    print(f"{contrasena}  ({fortaleza})")

print("\nPuedes copiar cualquiera de estas contrasenas.")
````
