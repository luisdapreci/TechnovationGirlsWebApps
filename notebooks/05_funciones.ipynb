## Definir una función

```python
def saludar():
    print("Hola!")
    print("Bienvenida a Technovation Girls.")

# Llamar la función
saludar()
saludar()
```

La palabra clave `def` le dice a Python que vas a definir una función. Esto **define** la función, pero no la ejecuta. Para ejecutarla, la **llamas** escribiendo su nombre con paréntesis.

## Funciones con parámetros

Los **parámetros** son variables que le pasas a una función:

```python
def saludar(nombre):
    print(f"Hola, {nombre}!")

saludar("Ana")
saludar("Lucia")
saludar("Valeria")
```

### Múltiples parámetros

```python
def presentar(nombre, edad, ciudad):
    print(f"Me llamo {nombre}, tengo {edad} anos y soy de {ciudad}.")

presentar("Sofia", 15, "Guadalajara")
presentar("Daniela", 13, "Bogota")
```

## Funciones que devuelven un valor

Muchas veces necesitas que una función **calcule algo y te devuelva el resultado**. Para eso usas `return`:

```python
def sumar(a, b):
    resultado = a + b
    return resultado

total = sumar(8, 5)
print(total)  # muestra: 13

print(sumar(100, 200))  # muestra: 300
```

### Diferencia entre `print` y `return`

Esta es una confusión común al principio:

```python
# Esta función MUESTRA el resultado pero no lo devuelve
def sumar_con_print(a, b):
    print(a + b)

# Esta función DEVUELVE el resultado para usarlo
def sumar_con_return(a, b):
    return a + b

resultado1 = sumar_con_print(3, 4)   # muestra 7 en pantalla
print(f"resultado1 = {resultado1}")   # None (no devolvió nada)

resultado2 = sumar_con_return(3, 4)  # no muestra nada
print(f"resultado2 = {resultado2}")   # 7
```

## Parámetros con valor por defecto

```python
def saludar(nombre, mensaje="Bienvenida"):
    print(f"{mensaje}, {nombre}!")

saludar("Ana")              # usa el valor por defecto
saludar("Lucia", "Hola")   # usa el valor que le pasamos
```

## Variables locales y globales

Las variables creadas **dentro** de una función son **locales**: solo existen dentro de la función.

```python
def calcular():
    resultado = 42  # variable local
    print(f"Dentro de la funcion: {resultado}")

calcular()

# Descomenta la siguiente línea para ver el error:
# print(resultado)  # NameError: resultado no existe aquí
```

Las variables creadas **fuera** de las funciones son **globales** y se pueden leer desde cualquier parte:

```python
nombre = "Sofia"  # variable global

def saludar():
    print(f"Hola, {nombre}")  # puede leer la variable global

saludar()
```

## Organizar el código con funciones

Las funciones sirven para **organizar** el programa en partes con responsabilidades claras:

```python
def obtener_datos():
    nombre = input("Como te llamas? ")
    edad = int(input("Cuantos anos tienes? "))
    return nombre, edad

def clasificar_edad(edad):
    if edad >= 18:
        return "adulta"
    else:
        return "menor de edad"

def mostrar_resultado(nombre, edad, categoria):
    print(f"Hola, {nombre}. Eres {categoria}.")
    print(f"En 10 anos tendras {edad + 10} anos.")

# Programa principal
nombre, edad = obtener_datos()
categoria = clasificar_edad(edad)
mostrar_resultado(nombre, edad, categoria)
```

## Devolver múltiples valores

```python
def minimo_y_maximo(numeros):
    return min(numeros), max(numeros)

menor, mayor = minimo_y_maximo([4, 1, 9, 2, 7])
print(f"Minimo: {menor}, Maximo: {mayor}")
```

## Ejemplo completo: calculadora de promedio

```python
def pedir_calificaciones():
    calificaciones = []
    print("Ingresa tus calificaciones (escribe 'fin' para terminar):")

    while True:
        entrada = input("Calificacion: ")
        if entrada.lower() == "fin":
            break
        calificaciones.append(float(entrada))

    return calificaciones

def calcular_promedio(calificaciones):
    if len(calificaciones) == 0:
        return 0
    return sum(calificaciones) / len(calificaciones)

def evaluar_promedio(promedio):
    if promedio >= 90:
        return "Excelente"
    elif promedio >= 75:
        return "Muy bien"
    elif promedio >= 60:
        return "Aprobado"
    else:
        return "Reprobado"

# Programa principal
calificaciones = pedir_calificaciones()
promedio = calcular_promedio(calificaciones)
evaluacion = evaluar_promedio(promedio)

print(f"\nCalificaciones: {calificaciones}")
print(f"Promedio: {promedio:.1f}")
print(f"Resultado: {evaluacion}")
```
