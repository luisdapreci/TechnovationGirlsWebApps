## Acceder a elementos por índice

Los índices **empiezan en 0**:

```python
frutas = ["manzana", "pera", "uva", "mango"]
#           0          1       2      3

print(frutas[0])  # manzana
print(frutas[2])  # uva
print(frutas[-1]) # mango (último)
print(frutas[-2]) # uva (penúltimo)
```

---

## Modificar elementos y métodos de listas

```python
frutas = ["manzana", "pera"]

frutas.append("uva")       # agrega al final
print(frutas)

frutas.insert(1, "mango")  # inserta en la posición 1
print(frutas)
```

```python
frutas = ["manzana", "pera", "uva", "mango"]

frutas.remove("pera")   # elimina por valor
print(frutas)

frutas.pop()            # elimina el último
print(frutas)
```

```python
numeros = [4, 1, 9, 2, 7, 1]

print(numeros.index(9))   # 2 (posición del valor 9)
print(numeros.count(1))   # 2 (cuántas veces aparece)

numeros.sort()
print(numeros)            # [1, 1, 2, 4, 7, 9]
```

```python
frutas = ["manzana", "pera", "uva"]

print(len(frutas))          # 3
print("pera" in frutas)     # True
print("mango" in frutas)    # False
```

## Recorrer una lista con `for`

```python
puntajes = [70, 85, 60, 92, 55]

for puntaje in puntajes:
    if puntaje >= 60:
        print(f"{puntaje} — Aprobado")
    else:
        print(f"{puntaje} — Reprobado")
```

Si necesitas el índice además del valor, usa `enumerate()`:

```python
participantes = ["Ana", "Lucia", "Valeria"]

for i, nombre in enumerate(participantes):
    print(f"{i + 1}. {nombre}")
```

---

## Sublistas con slicing

```python
numeros = [10, 20, 30, 40, 50, 60]

print(numeros[1:4])   # [20, 30, 40]
print(numeros[:3])    # [10, 20, 30]
print(numeros[3:])    # [40, 50, 60]
print(numeros[::2])   # [10, 30, 50] — de dos en dos
```

## Diccionarios

Un diccionario guarda pares de **clave: valor**. Se escribe entre llaves `{}`:

```python
estudiante = {
    "nombre": "Sofia",
    "edad": 15,
    "ciudad": "Monterrey",
    "promedio": 8.7
}

print(estudiante["nombre"])   # Sofia
print(estudiante["edad"])     # 15

# .get() evita errores si la clave no existe
print(estudiante.get("telefono", "N/A"))  # N/A
```

```python
# Modificar y agregar entradas
estudiante["edad"] = 16               # modificar
estudiante["escuela"] = "Tec de MTY"  # agregar nueva clave

print(estudiante)
```

---

## Recorrer un diccionario

```python
persona = {"nombre": "Ana", "edad": 14, "ciudad": "CDMX"}

# Claves y valores juntos
for clave, valor in persona.items():
    print(f"{clave}: {valor}")
```

```python
# Verificar si una clave existe
if "nombre" in persona:
    print("Tiene nombre registrado")

if "telefono" not in persona:
    print("No tiene telefono registrado")
```

## Listas de diccionarios

Una combinación muy común donde cada diccionario representa un elemento:

```python
equipo = [
    {"nombre": "Ana",     "rol": "Desarrolladora", "edad": 15},
    {"nombre": "Lucia",   "rol": "Disenadora",     "edad": 14},
    {"nombre": "Valeria", "rol": "Investigadora",  "edad": 16},
]

for integrante in equipo:
    print(f"{integrante['nombre']} — {integrante['rol']}")
```

## Ejemplo completo: agenda de contactos

```python
agenda = []

def agregar_contacto(nombre, telefono):
    contacto = {"nombre": nombre, "telefono": telefono}
    agenda.append(contacto)
    print(f"Contacto '{nombre}' agregado.")

def mostrar_agenda():
    if len(agenda) == 0:
        print("La agenda esta vacia.")
        return
    print("\n--- Agenda ---")
    for i, contacto in enumerate(agenda, 1):
        print(f"{i}. {contacto['nombre']} — {contacto['telefono']}")

# Programa principal
agregar_contacto("Ana Garcia", "555-1234")
agregar_contacto("Lucia Lopez", "555-5678")
agregar_contacto("Valeria Ruiz", "555-9012")

mostrar_agenda()
```

## Resumen

| Concepto | Qué aprendiste |
|---|---|
| Lista `[]` | Colección ordenada y mutable |
| Índice | Posición de un elemento; empieza en 0 |
| `.append()`, `.insert()` | Agregar elementos |
| `.remove()`, `.pop()` | Eliminar elementos |
| `.sort()`, `.reverse()` | Ordenar una lista |
| `enumerate()` | Recorrer con índice y valor |
| Slicing `[a:b]` | Obtener una sublista |
| Diccionario `{}` | Colección de pares clave-valor |
| `.get()` | Acceder sin riesgo de error |
| `.items()` | Recorrer claves y valores |
| Lista de diccionarios | Representar colecciones de objetos |
