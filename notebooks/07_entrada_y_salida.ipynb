## Abrir un archivo con `open()`

| Modo | Significado | Qué pasa si el archivo no existe |
|---|---|---|
| `"r"` | Leer *(read)* | Error `FileNotFoundError` |
| `"w"` | Escribir *(write)*, borra contenido anterior | Lo crea |
| `"a"` | Agregar al final *(append)* | Lo crea |

## La forma recomendada: `with`

Usar `with` es mejor práctica porque **cierra el archivo automáticamente**:

---

## Escribir archivos

````python
# Crear un archivo de texto
with open("notas.txt", "w") as archivo:
    archivo.write("Primera linea\n")
    archivo.write("Segunda linea\n")
    archivo.write("Tercera linea\n")

print("Archivo 'notas.txt' creado!")
````

---

## Leer archivos

````python
# Leer todo el contenido de una vez
with open("notas.txt", "r") as archivo:
    contenido = archivo.read()
    print(contenido)
````

````python
# Leer línea por línea
with open("notas.txt", "r") as archivo:
    for linea in archivo:
        print(linea.strip())  # .strip() elimina el salto de linea extra
````

---

## Agregar contenido sin borrar lo anterior

````python
with open("notas.txt", "a") as archivo:
    archivo.write("Esta linea se agrega al final\n")

# Verificar que se agregó
with open("notas.txt", "r") as archivo:
    print(archivo.read())
````

## Manejar el error cuando el archivo no existe

Si intentas leer un archivo que no existe, Python lanza un `FileNotFoundError`. Manéjalo con `try` y `except`:

````python
try:
    with open("archivo_que_no_existe.txt", "r") as archivo:
        contenido = archivo.read()
        print(contenido)
except FileNotFoundError:
    print("El archivo no existe todavia.")
````

## Verificar si un archivo existe

````python
import os

if os.path.exists("notas.txt"):
    print("El archivo 'notas.txt' existe!")
else:
    print("El archivo no existe.")
````

## Trabajar con archivos CSV

Los archivos **CSV** *(Comma-Separated Values)* son archivos de texto donde cada línea es un registro y los valores están separados por comas.

```
nombre,edad,ciudad
Ana,15,Monterrey
Lucia,14,Guadalajara
Valeria,16,CDMX
```

### Escribir un CSV

````python
import csv

participantes = [
    {"nombre": "Ana",     "edad": 15, "ciudad": "Monterrey"},
    {"nombre": "Lucia",   "edad": 14, "ciudad": "Guadalajara"},
    {"nombre": "Valeria", "edad": 16, "ciudad": "CDMX"},
]

with open("participantes.csv", "w", newline="") as archivo:
    campos = ["nombre", "edad", "ciudad"]
    escritor = csv.DictWriter(archivo, fieldnames=campos)

    escritor.writeheader()   # escribe la fila de encabezados
    escritor.writerows(participantes)

print("Archivo CSV guardado.")
````

### Leer un CSV

````python
import csv

with open("participantes.csv", "r") as archivo:
    lector = csv.DictReader(archivo)
    for fila in lector:
        print(f"{fila['nombre']} — {fila['ciudad']}")
````

## Ejemplo completo: diario personal

````python
import os
from datetime import date

ARCHIVO = "diario.txt"

def nueva_entrada():
    hoy = date.today().strftime("%d/%m/%Y")
    texto = input("Escribe tu entrada de hoy:\n> ")
    with open(ARCHIVO, "a") as archivo:
        archivo.write(f"\n[{hoy}]\n{texto}\n")
    print("Entrada guardada.")

def leer_diario():
    if not os.path.exists(ARCHIVO):
        print("Tu diario esta vacio. Escribe tu primera entrada.")
        return
    with open(ARCHIVO, "r") as archivo:
        print("\n--- Mi diario ---")
        print(archivo.read())

# Prueba el diario:
nueva_entrada()
leer_diario()
````

## Resumen

| Concepto | Qué aprendiste |
|---|---|
| `open(archivo, modo)` | Abre un archivo para leerlo o escribirlo |
| `with open(...) as f` | Forma segura que cierra automáticamente |
| `"r"`, `"w"`, `"a"` | Modos de lectura, escritura y agregar |
| `.read()` | Lee todo el contenido |
| `.write()` | Escribe un string en el archivo |
| `.strip()` | Elimina espacios y saltos de línea |
| `try / except` | Maneja errores sin detener el programa |
| Módulo `csv` | Lee y escribe archivos CSV |
| `os.path.exists()` | Verifica si un archivo existe |
