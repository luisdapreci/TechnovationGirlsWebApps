Imagina que quieres mostrar los números del 1 al 100, o saludar a cada participante de una lista. Hacerlo con `print` uno por uno sería muy lento y repetitivo.

Para eso existen los **ciclos**: estructuras que repiten un bloque de instrucciones automáticamente.

Python tiene dos tipos de ciclos: `for` y `while`.

---

## El ciclo `for`

Usa `for` cuando sabes de antemano **cuántas veces** quieres repetir algo.

### Recorrer un rango de números con `range()`

````python
for numero in range(5):
    print(numero)
````

`range(5)` genera los números del `0` al `4` (5 números en total). Para empezar desde 1:

````python
for numero in range(1, 6):
    print(numero)
````

Para contar de dos en dos (o cualquier otro paso):

````python
for numero in range(0, 11, 2):
    print(numero)
````

### Recorrer una lista de elementos

````python
participantes = ["Ana", "Lucia", "Valeria", "Sofia"]

for nombre in participantes:
    print(f"Bienvenida, {nombre}!")
````

### Recorrer un string carácter por carácter

````python
palabra = "Python"

for letra in palabra:
    print(letra)
````

## El ciclo `while`

Usa `while` cuando no sabes exactamente cuántas repeticiones necesitas, sino que quieres repetir **mientras una condición sea verdadera**.

````python
contador = 1

while contador <= 5:
    print(contador)
    contador = contador + 1
````

**Muy importante:** asegúrate de que la condición eventualmente se vuelva `False`. Si no, el ciclo corre para siempre (**ciclo infinito**). Si accidentalmente creas uno, puedes detenerlo con el botón de stop del notebook.

### Pedir datos hasta que sean válidos

````python
respuesta = ""

while respuesta != "si" and respuesta != "no":
    respuesta = input("Deseas continuar? (si/no) ").lower()

print(f"Elegiste: {respuesta}")
````

## Acumular valores dentro de un ciclo

Un patrón muy útil es usar una variable que **acumula** un resultado:

### Sumar una serie de números

````python
total = 0

for numero in range(1, 6):
    total = total + numero

print(f"La suma de 1 a 5 es: {total}")
````

### Contar cuántos elementos cumplen una condición

````python
puntajes = [85, 42, 91, 60, 73, 38, 88]
aprobados = 0

for puntaje in puntajes:
    if puntaje >= 60:
        aprobados = aprobados + 1

print(f"Aprobaron {aprobados} de {len(puntajes)} estudiantes")
````

## `break` y `continue`

### `break` — salir del ciclo antes de que termine

````python
for numero in range(1, 11):
    if numero == 6:
        break
    print(numero)
````

### `continue` — saltar al siguiente paso

````python
for numero in range(1, 11):
    if numero % 2 == 0:
        continue
    print(numero)
````

---

## `for` vs `while`: cuándo usar cada uno

| Situación | Ciclo recomendado |
|---|---|
| Sabes cuántas repeticiones necesitas | `for` |
| Recorres una lista o secuencia | `for` |
| Repites hasta que se cumpla una condición | `while` |
| Pides datos hasta que sean válidos | `while` |

---

## Ejemplo completo: juego de adivinanza

````python
import random

numero_secreto = random.randint(1, 10)
intentos = 0
adivinado = False

print("Adivina el numero entre 1 y 10.")

while not adivinado:
    intento = int(input("Tu numero: "))
    intentos = intentos + 1

    if intento < numero_secreto:
        print("Demasiado bajo, intenta de nuevo.")
    elif intento > numero_secreto:
        print("Demasiado alto, intenta de nuevo.")
    else:
        adivinado = True

print(f"Correcto! Lo adivinaste en {intentos} intento(s).")
````

## Resumen

| Concepto | Qué aprendiste |
|---|---|
| `for` | Repite un bloque un número definido de veces o recorre una colección |
| `while` | Repite un bloque mientras una condición sea verdadera |
| `range(inicio, fin, paso)` | Genera una secuencia de números |
| Variable acumuladora | Variable que se actualiza en cada repetición |
| `break` | Sale del ciclo inmediatamente |
| `continue` | Salta al siguiente paso del ciclo |
